## FizzBuzz
This is essentially the classic FizzBuzz problem — a drinking game turned interview question. You'd be surprised how many SWE candidates struggle with it. Let's make sure we're not one of them.

```python
def fizz_buzz(x, lower_bound=2, fizz=3, buzz=5):
	return [number for number in range(lower_bound, x + 1) if (number % fizz) == 0 or (number % buzz) == 0]
```

Passed your rigorous interview bar? Don't pop champagne yet — we have maths to do.

The problem deliberately stops at 10 to hide that multiples of 15 get counted twice (FizzBuzz!). There's a great [mCoding video](https://www.youtube.com/watch?v=Qgevy75co8c) on why the naive while-loop approach isn't Pythonic.

```python
def multiples_of_3s_and_5s_using_while_loop():
	"""UnPythonic approach — don't do this"""
	i = 0
	total = 0
	while i < 1000:
		if i % 3 == 0 or i % 5 == 0:
			total += i
		
		i += 1

	return total
```

## Pythonic Approach

In the spirit of [functional programming](https://stackoverflow.com/questions/1017621/why-isnt-python-very-good-for-functional-programming):

```python
# from solutions.latest.p0001 import _sigma_n_with_multiplier

def multiples_of_3s_and_5s_using_sum(upper_bound=1000):
	divisible_by_3_or_5 = lambda x: x % 3 == 0 or x % 5 == 0
	return sum(filter(divisible_by_3_or_5, range(upper_bound)))
```

## Gaussian Summation

Legend has it young Gauss was asked to sum 1 to 100 — expected to take all day. He noticed that 1+100 = 2+99 = ... = 101, with 50 such pairs:

$$\sum_{i=1}^{n} i = \frac{n(n+1)}{2}$$

How does this help? Multiples of 3 are:

$$3 + 6 + 9 + ... + 3n = 3(1 + 2 + ... + n) = 3 \cdot \sigma(n)$$

where $\sigma(n) = \frac{n(n+1)}{2}$.

**But wait** — multiples of 15 (like 15, 30, 45...) are counted in both the 3s and 5s. We need **inclusion-exclusion**:

$$\text{answer} = \sigma_3 + \sigma_5 - \sigma_{15}$$

So naively thinking we're done here is wishful thinking (and irresponsible coding!) as we can come to the conclusion that we're adding multiples of 15 twice. The problem purposefully avoids listing number until 15, so we can have some fun figuring this out (also note that it's partially implied by fizz-or-buzz).
