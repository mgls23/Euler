# [Problem 2: Even Fibonacci numbers](https://projecteuler.net/problem=2)


> Each new term in the Fibonacci sequence is generated by adding the previous two terms. Starting with 1 and 2, sum the even-valued terms not exceeding four million.


## Approach (derive and use the even-term recurrence)
Start with the Fibonacci rule: `F(n+1) = F(n) + F(n-1)`.

**Step 1: derive a 3-step relation**
``
F(n+2) = F(n+1) + F(n)
F(n+3) = F(n+2) + F(n+1)
       = (F(n+1) + F(n)) + F(n+1)
       = 2*F(n+1) + F(n)
``
So `F(n+3) = 2*F(n+1) + F(n)`.

**Step 2: apply it again to jump 6 steps**
``
F(n+6) = 2*F(n+4) + F(n+3)
F(n+4) = 2*F(n+2) + F(n+1)
F(n+5) = 2*F(n+3) + F(n+2)
``
Substituting and simplifying (or using standard Fibonacci identities) yields:
`F(n+6) = 4*F(n+3) + F(n)`

**Step 3: map to the even subsequence**
Every 3rd Fibonacci term is even, so define `E(k) = F(3k-1)`.
Then:
`E(k) = F(3k-1) = F((3(k-1)-1) + 6)`
`     = 4*F(3(k-1)-1) + F(3(k-2)-1)`
`     = 4*E(k-1) + E(k-2)`

This gives a tight recurrence for even terms: `E(k) = 4*E(k-1) + E(k-2)` with `E1=2`, `E2=8`.
We can now sum only even terms up to 4,000,000 in O(k) time and constant space.

### NFibonacciIterator: generic k-step skipping
The recurrence above shows that a fixed-step subsequence of Fibonacci numbers still follows a 2-term linear recurrence.
That is exactly what `NFibonacciIterator` encodes: it stores coefficients `(a, b)` so that
`next = a*prev + b*prev_prev` for the chosen step size `k`.

For `k=3` (even terms), the coefficients are `(a=4, b=1)` and the seeds are `[2, 8]`.
For other `k`, the coefficients come from the Fibonacci Q-matrix raised to the k-th power,
which guarantees the subsequence stays exact while skipping intermediate terms.
